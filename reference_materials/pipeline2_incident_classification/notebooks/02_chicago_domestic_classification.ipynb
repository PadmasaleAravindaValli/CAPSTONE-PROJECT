# Pipeline 2: Incident-Level Domestic Status Classification

This notebook implements the exact methodology from the capstone methodology paper (`final methology.docx`).

### Objectives:
1. Process 757,898 Chicago crime incidents (2023–2025).
2. Enforce strict leakage prevention (excluding `Description`, IUCR, timestamps, row identifiers).
3. Chronological split: Train (2023–Jun 2024), Validation (Jul–Dec 2024), Locked Test (2025).
4. Train 6 predictive models: Logistic Regression, Random Forest, HistGradientBoosting, XGBoost, MLP (Embeddings), FT-Transformer.
5. Select optimal decision thresholds on the validation set and evaluate on 2025 test records.
