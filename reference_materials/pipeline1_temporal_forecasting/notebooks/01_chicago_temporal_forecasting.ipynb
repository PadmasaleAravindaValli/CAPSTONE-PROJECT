# Pipeline 1: Spatio-Temporal Violent Crime Burden Forecasting

This notebook replicates the **Major-Project** methodology on the Chicago Crime dataset.

### Objectives:
1. Aggregate raw incident reports into Beat-Month panels (2023–2025).
2. Group crimes into 6 core offense categories.
3. Construct an ordered 5-level `TARGET_VIOLENT_LEVEL` target using frozen training quantiles.
4. Engineer 5 enhanced features including Shannon entropy (`CRIME_DIVERSITY_INDEX`) and year-over-year rate of change.
5. Compare observed-period baseline classification (Naive Bayes, Extra Trees, MLP) with **strict chronological sequence forecasting** (GRU, LSTM, and Tabular Logistic Regression).


```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

# Load aggregated data
data_path = Path('../../data/processed/pipeline1_temporal_forecasting/chicago_enhanced_features.csv')
df = pd.read_csv(data_path)
print(f'Total Beat-Month Records: {len(df):,}')
df.head()
```

## 1. Target Distribution and Quintile Boundaries
The target `TARGET_VIOLENT_LEVEL` is categorized into 5 ordered quintiles:
- Class 1: Lowest burden
- Class 2: Low-to-moderate burden
- Class 3: Moderate burden
- Class 4: Moderate-to-high burden
- Class 5: Highest burden


```python
plt.figure(figsize=(8, 4))
sns.countplot(data=df, x='TARGET_VIOLENT_LEVEL', palette='Blues_d')
plt.title('Distribution of Target Violent Crime Burden Levels')
plt.xlabel('Violent Crime Burden Level (1=Lowest, 5=Highest)')
plt.ylabel('Observation Count')
plt.show()
```
